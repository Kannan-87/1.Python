```python
# print Odd Numbers in the list

lists = [20,10,16,19,25,1,276,188]

for num in lists:
  if((num%2)==1):
    print(num, "is odd")
```

```python
# print Even numbers in the list

lists = [20,10,16,19,25,1,276,188]

for num in lists:
  if((num%2)!=1):
    print(num, "is even")
```
